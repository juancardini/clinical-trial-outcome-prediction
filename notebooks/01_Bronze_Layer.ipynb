Connection Details to AACT Database

```python
url = "jdbc:postgresql://aact-db.ctti-clinicaltrials.org:5432/aact"
user= "juanmartincardini"
password= "1161992"
driver="org.postgresql.Driver"
```

### Read Studies Data: ctgov.studies

```python

raw_df_studies=(spark.read
.format("jdbc")
.option("url",url)
.option("driver",driver)
.option("dbtable","ctgov.studies")
.option("user",user)
.option("password",password)
.load()
)
(raw_df_studies.write
 .format("delta")
.mode("overwrite")
.saveAsTable("bronze.raw_studies")
)
display(raw_df_studies.limit(5))
```
