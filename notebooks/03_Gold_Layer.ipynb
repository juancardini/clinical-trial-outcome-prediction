#Gold

###Studies Nodes

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_studies AS
SELECT DISTINCT
    study_id,
    brief_title,
    start_date,
    completion_date,
    study_type,
    clinical_phase as phase,
    overall_status,
    approved
FROM silver.studies_diabetes;
select   * from gold.nodes_studies limit 10
```

###Conditions

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_conditions AS
SELECT DISTINCT
    condition_id,
    condition_name_raw as condition_name,
    condition_category
FROM silver.conditions c
JOIN silver.studies_diabetes s
  ON c.study_id = s.study_id;  
  select * from gold.nodes_conditions limit 10
```

###Relationships Studies-Conditions

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_condition AS
SELECT DISTINCT
    s.study_id,
    c.condition_name_raw as condition_name,
    'HAS_CONDITION' AS rel_type
FROM silver.conditions c
JOIN silver.studies_diabetes s
  ON c.study_id = s.study_id;
select * from gold.rels_study_condition limit 10
```

###Countries Nodes

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_countries AS

    SELECT DISTINCT country_id,
        country_name
    FROM silver.countries c
    JOIN silver.studies_diabetes s
      ON c.study_id = s.study_id;
select * from gold.nodes_countries limit 10
```

###Relationships Countries-Studies

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_country AS
SELECT DISTINCT
    s.study_id,
    c.country_name,
    'DEVELOPED_IN_COUNTRY' AS rel_type
FROM silver.countries c
JOIN silver.studies_diabetes s
    ON c.study_id = s.study_id
;
select * from gold.rels_study_country limit 10
```

###Nodes Interventions

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_interventions AS
SELECT DISTINCT
    i.intervention_id,
    i.intervention_name,
    i.intervention_type
FROM silver.interventions i
JOIN silver.studies_diabetes s
  ON i.study_id = s.study_id;
  
SELECT * FROM gold.nodes_interventions limit 10;
```

###Relationships Studies Interventions

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_intervention AS
SELECT DISTINCT
    s.study_id,
    i.intervention_name,
    'HAS_INTERVENTION' AS rel_type
FROM silver.interventions i
JOIN silver.studies_diabetes s
  ON i.study_id = s.study_id;

SELECT * FROM gold.rels_study_intervention limit 10;
```

###Sponsors

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_sponsors AS
SELECT DISTINCT
    sp.sponsor_id,
    sp.sponsor_name,
    sp.agency_class,
    sp.lead_or_collaborator
FROM silver.sponsors sp
JOIN silver.studies_diabetes s
  ON sp.study_id = s.study_id;

SELECT * FROM gold.nodes_sponsors limit 10 ;
```

###Relationships Studies Sponsors

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_sponsor AS
SELECT DISTINCT
    s.study_id,
    sp.sponsor_name,
    'HAS_SPONSOR' AS rel_type
FROM silver.sponsors sp
JOIN silver.studies_diabetes s
  ON sp.study_id = s.study_id;

SELECT * FROM gold.rels_study_sponsor limit 10;
```

###Designs

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_designs AS
SELECT DISTINCT
    d.design_id,
    d.allocation,
    d.primary_purpose,
    d.intervention_model,
    d.masking,
    d.subject_masked,
    d.caregiver_masked,
    d.investigator_masked
FROM silver.designs d
JOIN silver.studies_diabetes s
  ON d.study_id = s.study_id;

SELECT * FROM gold.nodes_designs limit 10;
```

###Relationships Studies Designs

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_design AS
SELECT DISTINCT
    s.study_id,
    d.design_id,
    'HAS_DESIGN' AS rel_type
FROM silver.designs d
JOIN silver.studies_diabetes s
  ON d.study_id = s.study_id;

SELECT * FROM gold.rels_study_design limit 10;
```

###Facilities

```python
%sql
CREATE OR REPLACE TABLE gold.nodes_facilities AS
SELECT DISTINCT
    f.facility_id,
    f.facility_name,
    f.city,
    f.state,
    f.country,
    f.latitude,
    f.longitude
FROM silver.facilities f
JOIN silver.studies_diabetes s
  ON f.study_id = s.study_id;

SELECT* FROM gold.nodes_facilities limit 10;
```

###Relationships Studies Fcilities

```python
%sql
CREATE OR REPLACE TABLE gold.rels_study_facility AS
SELECT DISTINCT
    s.study_id,
    f.facility_id,
    'HAS_FACILITY' AS rel_type
FROM silver.facilities f
JOIN silver.studies_diabetes s
  ON f.study_id = s.study_id;

SELECT * FROM gold.rels_study_facility limit 10;
```
