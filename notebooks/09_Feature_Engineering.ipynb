```python
%sql
create or replace temp view n_conditions as 
select count(distinct condition_name) as n_conditions,study_id 
from gold.rels_study_condition 
group by study_id;
select * from n_conditions limit 10
```

```python
%sql
create or replace temp view n_countries as 
select study_id,count(distinct country_name) as n_countries 
from gold.rels_study_country
group by study_id;
select * from n_countries limit 10
```

```python
%sql
create or replace temp view n_interventions as 
select study_id,count (distinct intervention_name) as n_interventions 
from gold.rels_study_intervention
group by study_id;
select * from n_interventions limit 10
```

```python
%sql
create or replace temp view n_sponsors as SELECT 
  r.study_id,
  COUNT(DISTINCT CASE 
    WHEN UPPER(s.lead_or_collaborator) = 'LEAD' 
      THEN s.sponsor_name 
  END) AS n_sponsor_lead,
  COUNT(DISTINCT CASE 
    WHEN UPPER(s.lead_or_collaborator) = 'COLLABORATOR' 
      THEN s.sponsor_name 
  END) AS n_sponsor_collaborator
FROM gold.rels_study_sponsor r
JOIN gold.nodes_sponsors s 
  ON UPPER(TRIM(s.sponsor_name)) = UPPER(TRIM(r.sponsor_name))
GROUP BY r.study_id;
select * from n_sponsors limit 10
```

```python
%sql

create or replace temp view n_facilities as 
  SELECT
    r.study_id,
    count(distinct f.facility_id) as n_facilities
    
  FROM gold.rels_study_facility r
  JOIN gold.nodes_facilities f
    ON r.facility_id = f.facility_id
    group by r.study_id
    ;
    select * from n_facilities limit 10
```

```python
%sql


CREATE OR REPLACE TEMP VIEW design_features AS
SELECT
  r.study_id,

  d.allocation,
  d.primary_purpose,
  d.intervention_model,
  d.masking,
  CASE 
    WHEN d.masking IS NULL OR UPPER(d.masking) = 'NONE' THEN 0 
    ELSE 1 
  END AS is_blinded,

  -- Masked Roles
  COALESCE(CAST(d.subject_masked      AS INT), 0) AS subject_masked_int,
  COALESCE(CAST(d.caregiver_masked    AS INT), 0) AS caregiver_masked_int,
  COALESCE(CAST(d.investigator_masked AS INT), 0) AS investigator_masked_int,

  ( COALESCE(CAST(d.subject_masked      AS INT), 0)
  + COALESCE(CAST(d.caregiver_masked    AS INT), 0)
  + COALESCE(CAST(d.investigator_masked AS INT), 0)
  ) AS n_roles_masked

FROM gold.rels_study_design r
JOIN gold.nodes_designs d
  ON r.design_id = d.design_id;
select * from design_features limit 10
```

```python
%sql
CREATE OR REPLACE TEMP VIEW dataset_ml AS
SELECT
    g.study_id,
s.brief_title,
    -- Clinical Variables
    
    s.approved,

    -- Graph Metrics
    g.degree_total,
    g.degree_norm,
    g.pagerank_norm,
    g.betweenness_norm,
    g.closeness_norm,
    g.community_louvain,

  -- Quantitative Variables
    nc.n_conditions,
    nctr.n_countries,
    ni.n_interventions,
    nf.n_facilities,
    --EMBEDDINGS
    g.emb_0,
    g.emb_1,
    g.emb_2,
    g.emb_3,
    g.emb_4,
    g.emb_5,
    g.emb_6,
    g.emb_7,
    g.emb_8,
    g.emb_9,
    g.emb_10,
    g.emb_11,
    g.emb_12,
    g.emb_13,
    g.emb_14,
    g.emb_15,
    g.emb_16,
    g.emb_17,
    g.emb_18,
    g.emb_19,
    g.emb_20,
    g.emb_21,
    g.emb_22,
    g.emb_23,
    g.emb_24,
    g.emb_25,
    g.emb_26,
    g.emb_27,
    g.emb_28,
    g.emb_29,
    g.emb_30,
    g.emb_31,
    g.emb_32,
    g.emb_33,
    g.emb_34,
    g.emb_35,
    g.emb_36,
    g.emb_37,
    g.emb_38,
    g.emb_39,
    g.emb_40,
    g.emb_41,
    g.emb_42,
    g.emb_43,
    g.emb_44,
    g.emb_45,
    g.emb_46,
    g.emb_47,
    g.emb_48,
    g.emb_49,
    g.emb_50,
    g.emb_51,
    g.emb_52,
    g.emb_53,
    g.emb_54,
    g.emb_55,
    g.emb_56,
    g.emb_57,
    g.emb_58,
    g.emb_59,
    g.emb_60,
    g.emb_61,
    g.emb_62,
    g.emb_63,

    -- SPONSORS
    ns.n_sponsor_lead,
    ns.n_sponsor_collaborator,
    COALESCE(ns.n_sponsor_lead, 0) 
      + COALESCE(ns.n_sponsor_collaborator, 0) AS n_sponsors_total,

    -- Study Design
    d.allocation,
    d.primary_purpose,
    d.intervention_model,
    d.masking,
    d.is_blinded
FROM df_grafo g
LEFT JOIN gold.nodes_studies   s   ON s.study_id   = g.study_id
LEFT JOIN n_conditions         nc  ON nc.study_id  = g.study_id
LEFT JOIN n_countries          nctr ON nctr.study_id = g.study_id
LEFT JOIN n_interventions      ni  ON ni.study_id  = g.study_id
LEFT JOIN n_sponsors           ns  ON ns.study_id  = g.study_id
LEFT JOIN n_facilities         nf  ON nf.study_id  = g.study_id
LEFT JOIN design_features      d   ON d.study_id   = g.study_id;
select * from dataset_ml limit 20
```
